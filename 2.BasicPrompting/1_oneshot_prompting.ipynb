#### One shot prompting

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

example_input = "I absolutely loved the new restaurant downtown!"
example_output = """
<sentiment>
  <label>positive</label>
  <confidence>0.95</confidence>
</sentiment>"""

system_prompt = f"""You analyze sentiment and respond only in XML using this structure:
<sentiment>
  <label>positive|negative|neutral</label>
  <confidence>0.0-1.0</confidence>
</sentiment>

For any other query just say I cannot do it. 

Example input: 
{example_input}
Example output:
{example_output}
"""
user_query = "The delivery was late and the food arrived cold."
input = f"""
Input: {user_query}
Output: 
"""


```

```python
print(system_prompt)
```

```python
response = client.responses.create(
    model="gpt-4o-mini",
    instructions=system_prompt,
    input=input
)
print(response.output_text)
```

### Few Shot Prompting

```python
system_prompt = """You analyze sentiment and respond only in XML using this structure:
<sentiment>
  <label>positive|negative|neutral</label>
  <confidence>0.0-1.0</confidence>
</sentiment>"""

examples = [
    (
        "I absolutely loved the new restaurant downtown!",
        """<sentiment>
  <label>positive</label>
  <confidence>0.95</confidence>
</sentiment>""",
    ),
    (
        "The movie was okay, nothing special.",
        """<sentiment>
  <label>neutral</label>
  <confidence>0.80</confidence>
</sentiment>""",
    ),
    (
        "Worst customer service I have ever experienced.",
        """<sentiment>
  <label>negative</label>
  <confidence>0.97</confidence>
</sentiment>""",
    ),
]

user_query = "The hotel room was clean and the staff were friendly, but the wifi barely worked."
```

```python
few_shot_input = "\n\n".join(
    f"Example input: {example_input}\nExample output:\n{example_output}"
    for example_input, example_output in examples
)


print(few_shot_input)
```

```python
few_shot_input +=  f"\n\nInput: {user_query}\nOutput:"
# print(few_shot_input)
response = client.responses.create(
    model="gpt-4o-mini",
    instructions=system_prompt,
    input=few_shot_input
)
print(response.output_text)
```
