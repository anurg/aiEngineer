#### Zero Shot Prompting

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

system_prompt= """You are a helpful assistant. 
    Summarize the user's text clearly and concisely in 1 sentence, preserving the most important ideas."""
user_query= """
Artificial intelligence is changing how people work, learn, and communicate. Modern AI systems can analyze large amounts of data, 
recognize patterns, generate text and images, and help people solve complex problems more quickly. Businesses use these systems to 
automate repetitive tasks, improve customer support, and make better decisions, while students and researchers use them to explore ideas 
and understand difficult subjects. However, AI also creates concerns about privacy, inaccurate information, job displacement, and unfair bias. 
Because of these opportunities and risks, organizations must use AI responsibly, keep humans involved in important decisions, and regularly 
evaluate whether their systems are safe, accurate, and fair.
"""
response = client.responses.create(
    model="gpt-4o-mini",
    instructions=system_prompt,
    input=user_query
)
print (response.output_text)
```
