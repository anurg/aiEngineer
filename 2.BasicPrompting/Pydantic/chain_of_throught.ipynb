```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI()
```

```python
Q = ("Roger has 5 tennis balls. He buys 2 more cans of tennis balls. "
     "Each can has 3 tennis balls. How many tennis balls does he have now?")

# --- Failure demo: base model, Completions API ---
fail = client.completions.create(
    model="davinci-002",
    prompt=f"Q: {Q}\nA: The answer (just the number) is",
    max_tokens=5, temperature=0
)
print("davinci-002 direct:", fail.choices[0].text)
```

```python
success = client.completions.create(
    model="gpt-4o-mini",
    prompt=f"Q: {Q}\nA: The answer (just the number) is",
    max_tokens=5, temperature=0
)
print("gpt-4o-mini direct:", success.choices[0].text)
```
